# GWTC-5 Robustness and Interpretation Analysis

This notebook runs a separately labeled exploratory analysis of the
frozen 18-event confirmatory result.

It does not modify or overwrite anything in
`results/confirmatory_analysis`. The script verifies the confirmatory
input hashes before calculating leave-one-event-out checks, alternative
event summaries, cross-metric agreement, finite-sampling diagnostics,
JSD bin sensitivity, and descriptive Tier B/C extensions.

Exploratory p-values in this stage do not replace or expand the frozen
confirmatory hypothesis family.

```python
from pathlib import Path
import os
import runpy
import sys

PROJECT_ROOT = Path(r"C:\Users\tmark\GWTC5")

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(PROJECT_ROOT)

SCRIPT = (
    PROJECT_ROOT
    / "src"
    / "robustness"
    / "GWTC5_robustness_analysis.py"
)

if not SCRIPT.exists():
    raise FileNotFoundError(
        "Place GWTC5_robustness_analysis.py in:\n"
        f"{SCRIPT.parent}"
    )

os.environ["GWTC5_PROJECT_ROOT"] = str(PROJECT_ROOT)

# The first robustness run is protected from accidental overwriting.
sys.argv = [str(SCRIPT)]
runpy.run_path(str(SCRIPT), run_name="__main__")
```

Run this cell only after the analysis cell finishes successfully.
