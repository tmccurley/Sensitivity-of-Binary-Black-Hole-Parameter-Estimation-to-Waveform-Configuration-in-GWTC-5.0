# GWTC-5 Frozen Confirmatory Analysis

This notebook runs the locked 18-event XPHM–XPNR confirmatory analysis.

It verifies the final sample, computes all frozen posterior metrics,
generates the 100-half-split finite-sampling noise floor, evaluates
H1–H3, and exports reproducible tables, figures, hashes, and a log.

The analysis refuses to overwrite an existing result directory by
default.

```python
from pathlib import Path
import os
import runpy
import sys

PROJECT_ROOT = Path(r"C:\Users\tmark\GWTC5")

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(PROJECT_ROOT)

SCRIPT = (
    PROJECT_ROOT
    / "src"
    / "confirmatory"
    / "GWTC5_confirmatory_analysis.py"
)

if not SCRIPT.exists():
    raise FileNotFoundError(
        "Place GWTC5_confirmatory_analysis.py in:\n"
        f"{SCRIPT.parent}"
    )

os.environ["GWTC5_PROJECT_ROOT"] = str(PROJECT_ROOT)

# No --overwrite flag is used. This protects the first confirmatory run.
sys.argv = [str(SCRIPT)]
runpy.run_path(str(SCRIPT), run_name="__main__")
```

Run this only after the analysis cell finishes successfully.
